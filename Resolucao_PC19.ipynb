## 2. Demonstrando o Corpus: Mac-Morpho

O *Mac-Morpho* é um corpus construído a partir de textos da Folha de São Paulo de 1994. Ele é amplamente utilizado para treinar marcadores (taggers) de *Part-of-Speech* (POS) em português do Brasil.

Vamos carregá-lo e demonstrar seu uso extraindo palavras, sentenças e suas respectivas anotações (tags).

```python
from nltk.corpus import mac_morpho

print("### Demonstração: Mac-Morpho ###\n")

# 1. Contagem básica
print(f"Total de palavras no corpus: {len(mac_morpho.words())}")
print(f"Total de sentenças no corpus: {len(mac_morpho.sents())}\n")

# 2. Acessando palavras anotadas (Palavra, Tag POS)
palavras_anotadas = mac_morpho.tagged_words()
print("Primeiras 10 palavras anotadas:")
print(palavras_anotadas[:10])
print("\n")

# 3. Acessando sentenças completas já anotadas
sentencas_anotadas = mac_morpho.tagged_sents()
print("Primeira sentença anotada:")
print(sentencas_anotadas[0])
```

### Exemplo Prático com Mac-Morpho
Vamos descobrir quais são as 5 tags morfológicas mais comuns neste corpus.

```python
from nltk.probability import FreqDist

# Extraímos apenas as tags da lista de tuplas (palavra, tag)
todas_as_tags = [tag for (palavra, tag) in mac_morpho.tagged_words()]

# Calculamos a distribuição de frequência
freq_tags = FreqDist(todas_as_tags)

print("As 5 classes gramaticais (tags) mais frequentes no Mac-Morpho:")
for tag, contagem in freq_tags.most_common(5):
    print(f"Tag: {tag:4} | Ocorrências: {contagem}")
    
# Nota:
# N = Substantivo (Noun)
# PREP = Preposição
# ART = Artigo
# V = Verbo
# PU = Pontuação
```

## 3. Demonstrando o Corpus: Floresta Sintática

A *Floresta Sintática* não apenas possui anotação de classes gramaticais (POS), mas é um *Treebank* completo, o que significa que anota a estrutura sintática (análise em árvore) das sentenças.

É mantido pelo projeto Linguateca.

```python
from nltk.corpus import floresta

print("### Demonstração: Floresta Sintática ###\n")

# A Floresta possui tags próprias, que preservamos para não perder
# informações específicas da anotação sintática em português.
palavras_floresta = floresta.tagged_words()
print("Primeiras 10 palavras com as tags nativas:")
print(palavras_floresta[:10])
print("\n")

# Como é um Treebank, podemos acessar a árvore sintática gerada.
arvores = floresta.parsed_sents()

print("Estrutura da primeira árvore sintática:")
primeira_arvore = arvores[0]
print(primeira_arvore)

# Em um ambiente com suporte gráfico, primeira_arvore.draw()
# pode abrir uma janela com a árvore desenhada.
```
